#### **Question 1**: <b>Which states have the highest rates of a particular type of crime. Which economic indicators justify the reason for this?</b>

Chart Types that are suited to answer this question: 
* Chart Type 1: Stacked Bar Chart
* Chart Type 2: Grouped Bar Chart

The chart below is used to aid my next 2 visualizations , it is used to find the states with the least and highest rates for each crime.

```python

# --- Aggregate data at state level ---
avg_state = df.groupby(['State', 'Location', 'State Abbr']).agg(
    larceny      = ('Crime Rate_larceny',              'mean'),
    burglary     = ('Crime Rate_burglary',             'mean'),
    mvt          = ('Crime Rate_motor_vehicle_theft',  'mean'),
    robbery      = ('Crime Rate_robbery',              'mean'),
    poverty      = ('Poverty rate',                    'mean'),
    unemployment = ('Unemployment rate',               'mean'),
    gdp          = ('GDP per capita',                  'mean'),
    gini         = ('Gini coefficient',                'mean'),
    population   = ('Population',                      'mean'),
).reset_index()

# --- Find highest AND lowest state for each crime type ---
crime_cols = {
    'larceny': 'Larceny',
    'burglary': 'Burglary',
    'mvt': 'Motor Vehicle Theft',
    'robbery': 'Robbery'
}

rows = []
for col, name in crime_cols.items():
    for rank, func in [('Highest', 'idxmax'), ('Lowest', 'idxmin')]:
        idx = getattr(avg_state[col], func)()
        rows.append({
            'Crime Type': name,
            'Rank':       rank,
            'State':      avg_state.loc[idx, 'State'],
            'Rate':       avg_state.loc[idx, col],
            'Poverty':    avg_state.loc[idx, 'poverty'],
            'Unemployment': avg_state.loc[idx, 'unemployment'],
            'GDP':        avg_state.loc[idx, 'gdp'],
            'Gini':       avg_state.loc[idx, 'gini']
        })

highest_lowest_crime_states = pd.DataFrame(rows)

# --- Visualization ---
chart = alt.Chart(highest_lowest_crime_states).mark_bar().encode(
    x=alt.X('Rate:Q', title='Crime Rate (per 100,000 people)'),
    y=alt.Y('Crime Type:N', sort='-x'),
    xOffset=alt.XOffset('Rank:N'),          # groups Highest/Lowest side by side
    color=alt.Color(
        'Rank:N',
        scale=alt.Scale(domain=['Highest', 'Lowest'], range=['#d62728', '#1f77b4']),
        title='Rank'
    ),
    tooltip=[
        alt.Tooltip('Crime Type:N'),
        alt.Tooltip('Rank:N'),
        alt.Tooltip('State:N'),
        alt.Tooltip('Rate:Q',         format='.1f'),
        alt.Tooltip('Poverty:Q',      format='.1f'),
        alt.Tooltip('Unemployment:Q', format='.1f'),
        alt.Tooltip('GDP:Q',          format=',.0f'),
        alt.Tooltip('Gini:Q',         format='.3f')
    ]
).properties(
    title='States with Highest & Lowest Rate per Crime Type (2005–2015)',
    width=550,
    height=320
)

chart
```

```python


# --- Crime pair definitions (from observed findings) ---
crime_pairs = [
    {'Crime Type': 'Larceny',             'Rank': 'Highest', 'State': 'California'},
    {'Crime Type': 'Larceny',             'Rank': 'Lowest',  'State': 'South Dakota'},
    {'Crime Type': 'Burglary',            'Rank': 'Highest', 'State': 'Arkansas'},
    {'Crime Type': 'Burglary',            'Rank': 'Lowest',  'State': 'New York'},
    {'Crime Type': 'Motor Vehicle Theft', 'Rank': 'Highest', 'State': 'Nevada'},
    {'Crime Type': 'Motor Vehicle Theft', 'Rank': 'Lowest',  'State': 'Vermont'},
    {'Crime Type': 'Robbery',             'Rank': 'Highest', 'State': 'Nevada'},
    {'Crime Type': 'Robbery',             'Rank': 'Lowest',  'State': 'Wyoming'},
]
pair_df = pd.DataFrame(crime_pairs)

# --- Aggregate socioeconomic indicators for relevant states ---
states = pair_df['State'].unique().tolist()

avg_state = (
    df[df['State'].isin(states)]
    .groupby('State').agg(
        gdp_per_capita          = ('GDP per capita',                                                          'mean'),
        gini         = ('Gini coefficient',                                                        'mean'),
        corruption   = ('Number of corruption convictions per 1,000,000 people',                   'mean'),
        below_9_years_of_education    = ('People with less than 9 years of education/people with college degree or above', 'mean'),
        poverty      = ('Poverty rate',                                                            'mean'),
        unemployment = ('Unemployment rate',                                                       'mean'),
    ).reset_index()
)

# --- Melt and merge with pair info ---
# Nevada intentionally appears in both MVT and Robbery groups
df_viz = (
    avg_state
    .melt(id_vars='State', var_name='Indicator', value_name='Value')
    .merge(pair_df, on='State', how='left')
)

# --- Dropdown ---
indicators = avg_state.columns.drop('State').tolist()
dropdown  = alt.binding_select(options=indicators, name='Select Indicator: ')
selection = alt.selection_point(fields=['Indicator'], bind=dropdown, value=[{'Indicator': 'gdp'}])

# --- Chart ---
chart = alt.Chart(df_viz).mark_bar().encode(
    x=alt.X('Crime Type:N',
            title=None,
            sort=['Larceny', 'Burglary', 'Motor Vehicle Theft', 'Robbery'],
            axis=alt.Axis(labelAngle=-12, labelFontSize=12)),
    xOffset=alt.XOffset('Rank:N', sort=['Highest', 'Lowest']),
    y=alt.Y('Value:Q', title='Value'),
    color=alt.Color(
        'Crime Type:N',
        scale=alt.Scale(
            domain=['Larceny', 'Burglary', 'Motor Vehicle Theft', 'Robbery'],
            range=['#4C78A8', '#F58518', '#54A24B', '#E45756']
        ),
        title='Crime Type Pair'
    ),
    # Full opacity = Highest, dimmed = Lowest — easy visual contrast within each pair
    opacity=alt.condition(
        alt.datum.Rank == 'Highest',
        alt.value(1.0),
        alt.value(0.4)
    ),
    tooltip=[
        alt.Tooltip('Crime Type:N',  title='Crime Type'),
        alt.Tooltip('Rank:N',        title='Rank'),
        alt.Tooltip('State:N',       title='State'),
        alt.Tooltip('Indicator:N',   title='Indicator'),
        alt.Tooltip('Value:Q',       title='Value', format=',.2f'),
    ]
).add_params(
    selection
).transform_filter(
    selection
).properties(
    title='Socioeconomic Indicators — Highest vs Lowest Crime States by Crime Type',
    width=560,
    height=380
)

chart
```

```python

crime_pairs = [
    {'Crime Type': 'Larceny',             'Rank': 'Highest', 'State': 'California'},
    {'Crime Type': 'Larceny',             'Rank': 'Lowest',  'State': 'South Dakota'},
    {'Crime Type': 'Burglary',            'Rank': 'Highest', 'State': 'Arkansas'},
    {'Crime Type': 'Burglary',            'Rank': 'Lowest',  'State': 'New York'},
    {'Crime Type': 'Motor Vehicle Theft', 'Rank': 'Highest', 'State': 'Nevada'},
    {'Crime Type': 'Motor Vehicle Theft', 'Rank': 'Lowest',  'State': 'Vermont'},
    {'Crime Type': 'Robbery',             'Rank': 'Highest', 'State': 'Nevada'},
    {'Crime Type': 'Robbery',             'Rank': 'Lowest',  'State': 'Wyoming'},
]
pair_df = pd.DataFrame(crime_pairs)

# --- Aggregate ---
states = pair_df['State'].unique().tolist()
avg_state = (
    df[df['State'].isin(states)]
    .groupby('State').agg(
        gdp_per_capita             = ('GDP per capita',                                                                 'mean'),
        gini                       = ('Gini coefficient',                                                               'mean'),
        corruption                 = ('Number of corruption convictions per 1,000,000 people',                          'mean'),
        below_9_years_of_education = ('People with less than 9 years of education/people with college degree or above', 'mean'),
        poverty                    = ('Poverty rate',                                                                   'mean'),
        unemployment               = ('Unemployment rate',                                                              'mean'),
    ).reset_index()
)

df_viz = (
    avg_state
    .melt(id_vars='State', var_name='Indicator', value_name='Value')
    .merge(pair_df, on='State', how='left')
)

crime_order = ['Larceny', 'Burglary', 'Motor Vehicle Theft', 'Robbery']

# Dropdown 
indicators = avg_state.columns.drop('State').tolist()
dropdown  = alt.binding_select(options=indicators, name='Select Indicator: ')
selection = alt.selection_point(fields=['Indicator'], bind=dropdown, value=[{'Indicator': 'gdp_per_capita'}])

base = alt.Chart(df_viz).transform_filter(selection)

# Layer 1 
lines = base.mark_line(strokeWidth=2.5, color='#aaaaaa').encode(
    y=alt.Y('Crime Type:N', sort=crime_order, title=None, axis=alt.Axis(labelFontSize=12)),
    x=alt.X('Value:Q', title='Value'),
    detail='Crime Type:N',
    tooltip=alt.value(None)
)

# Layer 2 
dots = base.mark_circle(size=160).encode(
    y=alt.Y('Crime Type:N', sort=crime_order),
    x=alt.X('Value:Q'),
    color=alt.Color(
        'Rank:N',
        scale=alt.Scale(domain=['Highest', 'Lowest'], range=['#d62728', '#1f77b4']),
        title='Rank'
    ),
    tooltip=[
        alt.Tooltip('Crime Type:N', title='Crime Type'),
        alt.Tooltip('Rank:N',       title='Rank'),
        alt.Tooltip('State:N',      title='State'),
        alt.Tooltip('Indicator:N',  title='Indicator'),
        alt.Tooltip('Value:Q',      title='Value', format=',.2f'),
    ]
)

# Layer 3 the labels
labels_high = base.transform_filter(alt.datum.Rank == 'Highest').mark_text(
    fontSize=10, fontWeight='bold', dy=-10, dx=-30
).encode(
    y=alt.Y('Crime Type:N', sort=crime_order),
    x=alt.X('Value:Q'),
    text=alt.Text('State:N'),
    color=alt.value('#d62728')
)

labels_low = base.transform_filter(alt.datum.Rank == 'Lowest').mark_text(
    fontSize=10, fontWeight='bold', dy=-10, dx=30
).encode(
    y=alt.Y('Crime Type:N', sort=crime_order),
    x=alt.X('Value:Q'),
    text=alt.Text('State:N'),
    color=alt.value('#1f77b4')
)

# add_params once on the final layered chart
chart = (lines + dots + labels_high + labels_low).add_params(
    selection
).properties(
    title='Socioeconomic Indicators — Highest vs Lowest Crime States (Dumbbell Chart)',
    width=580,
    height=300
).configure_view(
    strokeWidth=0
).configure_axis(
    grid=True,
    gridColor='#eeeeee'
)

chart.interactive()
```

#### <b>**Question 2**: Which type of crime is the least and most associated with unemployment rate ? </b><br><b>And which type of crime is the least and most associated with the gini-coefficient (economic metric used to measure inequality) ?</b>

Chart Types that are suited to answer this question: 
* Chart Type 1: Correlation Heatmap
* Chart Type 2: Scatter Plot

```python
avg_state = df.groupby('State').agg(
    larceny=('Crime Rate_larceny','mean'),
    burglary=('Crime Rate_burglary','mean'),
    mvt=('Crime Rate_motor_vehicle_theft','mean'),
    robbery=('Crime Rate_robbery','mean'),
    unemployment=('Unemployment rate','mean'),
    gini=('Gini coefficient','mean')
).reset_index()

crime_cols = ['larceny','burglary','mvt','robbery']

corr_data = []
for col in crime_cols:
    corr_data.append({
        'Crime Type': col,
        'Unemployment': avg_state[col].corr(avg_state['unemployment']),
        'Gini': avg_state[col].corr(avg_state['gini'])
    })

corr_df = pd.DataFrame(corr_data).melt(
    id_vars='Crime Type',
    var_name='Metric',
    value_name='Correlation'
)

df_long = avg_state.melt(
    id_vars=['State', 'unemployment', 'gini'],
    value_vars=crime_cols,
    var_name='Crime Type',
    value_name='Crime Rate'
)

# --- Shared selection: click a cell in the heatmap ---
crime_select = alt.selection_point(fields=['Crime Type'], name='crime_sel')

# --- Chart 1: Heatmap ---
heatmap = alt.Chart(corr_df).mark_rect().encode(
    x=alt.X('Metric:N', title='Socioeconomic Metric'),
    y=alt.Y('Crime Type:N', title='Crime Type'),
    color=alt.Color('Correlation:Q',
                    scale=alt.Scale(scheme='redblue', domain=[-1, 1]),
                    title='Correlation'),
    opacity=alt.condition(crime_select, alt.value(1.0), alt.value(0.3)),
    tooltip=['Crime Type', 'Metric', alt.Tooltip('Correlation:Q', format='.2f')]
).add_params(
    crime_select
).properties(
    title='Crime–Socioeconomic Correlation Heatmap (click to filter)',
    width=300,
    height=250
)

text = heatmap.mark_text(baseline='middle').encode(
    text=alt.Text('Correlation:Q', format='.2f'),
    color=alt.condition(
        'abs(datum.Correlation) > 0.5',
        alt.value('white'),
        alt.value('black')
    ),
    opacity=alt.condition(crime_select, alt.value(1.0), alt.value(0.3))
)

# --- Chart 2: Scatter Unemployment ---
scatter_unemp = alt.Chart(df_long).mark_circle(size=80).encode(
    x=alt.X('unemployment:Q', title='Unemployment Rate'),
    y=alt.Y('Crime Rate:Q', title='Crime Rate'),
    color=alt.condition(crime_select, alt.Color('Crime Type:N', legend=None), alt.value('lightgray')),
    tooltip=['State', 'Crime Type', 'Crime Rate', 'unemployment']
).transform_filter(
    crime_select
).properties(
    title='Crime vs Unemployment'
)

# --- Chart 3: Scatter Gini ---
scatter_gini = alt.Chart(df_long).mark_circle(size=80).encode(
    x=alt.X('gini:Q', title='Gini Coefficient'),
    y=alt.Y('Crime Rate:Q', title='Crime Rate'),
    color=alt.condition(crime_select, alt.Color('Crime Type:N', legend=None), alt.value('lightgray')),
    tooltip=['State', 'Crime Type', 'Crime Rate', 'gini']
).transform_filter(
    crime_select
).properties(
    title='Crime vs Gini'
)

(heatmap + text) & (scatter_unemp | scatter_gini)
```

<b>a) What insights can you gain from these charts? </b>

    Larceny is the least correlated with the unemployment rate since it has the least correlation coefficient with the unemployment rate.
    Robbery is the most correlated with the unemployment rate since it has the highest correlation coefficient with the unemployment rate.

    
    Larceny is the least correlated with the Gini Coefficient since it has the least correlation coefficient with the Gini Coefficient.
    Robbery is the most correlated with the Gini Coefficient since it has the highest correlation coefficien with the Gini Coefficient.

#### <b> Question 3 : Does economic growth (GDP per capita increase) lead to lower crime rates, or is it offset by rising inequality ?</b>

Chart Types that are suited to answer this question:  
* Chart Type 1: Scatter Chart
* Chart Type 2: Correlation Heatmap 

```python
scatter = alt.Chart(df_state).mark_point(filled=True, opacity=0.8, size=150).encode(
    x=alt.X('gdp_change:Q', title='GDP per Capita Change'),
    y=alt.Y('gini_change:Q', title='Gini Coefficient Change'),
    shape=alt.Shape('crime_level:O',
                    title='Avg Crime Level',
                    scale=alt.Scale(range=['circle', 'square', 'triangle-up'])),
    color=alt.Color('group:N', title='Economic Scenario'),
    tooltip=[
        alt.Tooltip('State:N'),
        alt.Tooltip('gdp_change:Q', format='.0f', title='GDP Change'),
        alt.Tooltip('gini_change:Q', format='.3f', title='Gini Change'),
        alt.Tooltip('avg_crime:Q', format='.1f', title='Avg Crime'),
        alt.Tooltip('crime_level:O', title='Crime Level')
    ]
).transform_calculate(
    crime_level="datum.avg_crime < 2000 ? 'Low' : datum.avg_crime < 4000 ? 'Medium' : 'High'"
).properties(
    title='Crime Rate by Economic Trajectory (shape = avg crime level)',
    width=500,
    height=400
).interactive()

scatter
```
